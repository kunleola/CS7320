## GPT 2

Model Card: https://huggingface.co/openai-community/gpt2

A small 124M parameters pretrained model without fine tuning.

```python
from transformers import pipeline, set_seed
generator = pipeline('text-generation', model='gpt2')
set_seed(42)
answer = generator("18 plus 7 times 12 is", max_new_tokens=10, num_return_sequences=1)

print(answer)
print(answer[0]['generated_text'])
```

This model is not very good with math!

```python
from transformers import pipeline, set_seed
generator = pipeline('text-generation', model='gpt2')
set_seed(42)
answer = generator("What is 18 plus 7 times 12?", max_new_tokens=50, num_return_sequences=1)

print(answer)
print(answer[0]['generated_text'])
```

```python
from transformers import pipeline, set_seed
generator = pipeline('text-generation', model='gpt2')
set_seed(42)
answer = generator("What is 18 plus 7 times 12? Think step by step.", max_new_tokens=50, num_return_sequences=1)
print(answer)
print(answer[0]['generated_text'])
```

## Qwen 2.5: A Smarter Reasoning Model

Model card: https://huggingface.co/Qwen/Qwen2.5-1.5B


Number of Parameters: 1.54B

Qwen2.5 brings the following improvements upon Qwen2:

* Significantly more knowledge and has greatly improved capabilities in coding and mathematics, thanks to our specialized expert models in these domains.
* Significant improvements in instruction following, generating long texts (over 8K tokens), understanding structured data (e.g, tables), and generating structured outputs especially JSON. More resilient to the diversity of system prompts, enhancing role-play implementation and condition-setting for chatbots.
* Long-context Support up to 128K tokens and can generate up to 8K tokens.
* Multilingual support for over 29 languages, including Chinese, English, French, Spanish, Portuguese, German, Italian, Russian, Japanese, Korean, Vietnamese, Thai, Arabic, and more.

```python
from transformers import pipeline, set_seed
generator = pipeline('text-generation', model='Qwen/Qwen2.5-1.5B')
set_seed(42)
answer = generator("How much is 18 plus 7 times 12?", max_new_tokens=200, num_return_sequences=1)

print(answer)
print(answer[0]['generated_text'])
```

This model was already pretrained on reasoning tasks.
