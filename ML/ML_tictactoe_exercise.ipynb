## Data

I have used a script to create data for two random players playing tic-tac-toe and saved boards after each more x made to the following file:

https://github.com/mhahsler/CS7320-AI/blob/master/ML/tictactoe_data.csv

The first 9 columns are the tic-tac-toe board in the following layout:

```
0	1	2
3	4	5
6	7	8
```

The values mean: 0 = empty, 1 = player x, -1 player o

The last column y is the game outcome: 0 = draw, 1 = x wins, -1 y wins

We can learn with this type of data what boards are more likely to lead to a win and therefore what boards are more desirable. The idea is that the agent will choose actions to get to more desirable boards.

Read the csv file into a Pandas dataframe and separate the data into the board X and the response vector y.

````python
import pandas as pd
import numpy as np

# original GitHub blob URL (or you can provide the raw.githubusercontent.com URL)
url = "https://raw.githubusercontent.com/mhahsler/CS7320-AI/master/ML/tictactoe_data.csv"

# read CSV (no header in file)
df = pd.read_csv(url)

X = df.iloc[:, :-1]  # features (all columns except the last one)
y = np.array(df.iloc[:, -1])   # target (last column)

print("X:\n", X)
print("y:\n", y) 
````

Here is a function to look at a tic-tac-toe board. We look at the 4th board in the data set.

````python
def show_board(x):
    board = np.array(x, dtype=str).reshape(3, 3)
    board[board == '1'] = 'x'
    board[board == '-1'] = 'o'
    board[board == '0'] = ' '

    print(" | ".join(board[0,:]))
    print("-" * 9)
    print(" | ".join(board[1,:]))
    print("-" * 9)
    print(" | ".join(board[2,:]))
 
board = X.iloc[4,:]
print("Board:", board)
        
show_board(board)
````
